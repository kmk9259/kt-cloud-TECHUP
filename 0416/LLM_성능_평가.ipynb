# LLM의 성능 평가

LLM(Large Language Model)의 성능 평가는 단순히 “답을 잘하느냐”만 보는 것이 아니라,  
**언어모델 자체의 예측 능력**, **다운스트림 작업 성능**, **생성 품질**, **추론 효율**, **안전성/정렬성**까지 함께 보는 과정

LLM 평가는 보통 다음 여러 축으로 나뉨

- 언어모델링 성능
- 태스크별 정확도
- 생성 품질
- 추론/이해/코드 능력
- 효율성
- 안정성 및 안전성

---

# 1. 언어모델의 기본 목적과 평가

LLM은 기본적으로 다음 조건부 확률을 모델링.

$$
P(x_1, x_2, \dots, x_T)
= \prod_{t=1}^{T} P(x_t \mid x_1, \dots, x_{t-1})
$$

앞의 토큰들이 주어졌을 때 다음 토큰의 확률을 잘 예측하는 것이 기본 목표.

이 관점에서 가장 기본적인 평가는 다음과 같음.

- Negative Log-Likelihood (NLL)
- Cross-Entropy Loss
- Perplexity

---

# 2. Cross-Entropy Loss

가장 기본적인 학습 및 평가 지표.

정답 토큰 시퀀스가 $y_1, y_2, \dots, y_T$ 일 때, 언어모델의 평균 cross-entropy loss는 보통 다음처럼 씀.

$$
\mathcal{L}
=
-\frac{1}{T}\sum_{t=1}^{T}\log P(y_t \mid y_{<t})
$$

- $T$: 전체 토큰 수
- $y_t$: 시점 $t$의 정답 토큰
- $y_{<t}$: 그 이전까지의 정답 토큰

입니다.

이 값이 작을수록 모델이 정답 토큰에 높은 확률을 주고 있다는 뜻.

---

# 3. Perplexity (PPL)

언어모델 평가에서 매우 자주 등장하는 지표.

Perplexity는 cross-entropy의 지수 형태로 정의됨.

$$
\mathrm{PPL}
=
\exp\left(
-\frac{1}{T}\sum_{t=1}^{T}\log P(y_t \mid y_{<t})
\right)
$$

또는 loss를 $\mathcal{L}$ 라고 두면

$$
\mathrm{PPL} = e^{\mathcal{L}}
$$

## 해석

- PPL이 낮을수록 좋음
- 모델이 다음 토큰을 덜 헷갈린다는 뜻
- 언어모델 자체의 예측 능력을 반영

예를 들어

- PPL = 10
- PPL = 20

이라면 보통 PPL=10인 모델이 다음 토큰 예측을 더 잘하는 모델.

## 한계

Perplexity는 유용하지만 다음 한계가 있음.

- 실제 대화 품질과 완전히 일치하지 않음
- instruction-following 능력을 직접 반영하지 않음
- reasoning 품질을 직접 평가하지 않음
- 토크나이저가 다르면 직접 비교가 어렵기도 함

PPL은 **기초 언어모델링 능력**에는 좋지만, LLM의 실제 활용 성능 전체를 대표하지는 않음.

---

# 4. Token-level Accuracy

다음 토큰 예측 관점에서 정답 토큰을 정확히 맞추는 비율을 볼 수도 있음.

$$
\mathrm{Accuracy}
=
\frac{1}{T}\sum_{t=1}^{T}\mathbf{1}(\hat{y}_t = y_t)
$$

- $\hat{y}_t$: 모델이 예측한 토큰
- $y_t$: 실제 정답 토큰
- $\mathbf{1}(\cdot)$: 조건이 참이면 1, 아니면 0


다만 자연어 생성에서는 정답 표현이 하나만 있는 것이 아니므로, token-level accuracy는 생성형 태스크 전체 평가에는 제한적.

---

# 5. 다운스트림 태스크 기반 평가

LLM은 실제로 다양한 태스크에 사용되므로, 실전에서는 태스크별 성능 평가가 매우 중요.

예를 들어 다음과 같은 태스크들이 있습니다.

- 분류
- 요약
- 번역
- 질의응답
- 상식 추론
- 수학 문제 해결
- 코드 생성

이 경우에는 태스크 성격에 맞는 평가 지표를 사용.

---

# 6. 분류 태스크 평가

감성분석, 주제분류, NLI 같은 작업에서는 보통 다음 지표를 씁니다.

## (1) Accuracy

$$
\mathrm{Accuracy}
=
\frac{\text{정확히 맞춘 샘플 수}}{\text{전체 샘플 수}}
$$

또는

$$
\mathrm{Accuracy}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(\hat{y}_i = y_i)
$$

## (2) Precision

$$
\mathrm{Precision}
=
\frac{TP}{TP + FP}
$$

## (3) Recall

$$
\mathrm{Recall}
=
\frac{TP}{TP + FN}
$$

## (4) F1-score

$$
F1
=
\frac{2 \cdot \mathrm{Precision} \cdot \mathrm{Recall}}
{\mathrm{Precision} + \mathrm{Recall}}
$$


- $TP$: True Positive
- $FP$: False Positive
- $FN$: False Negative

불균형 데이터셋에서는 accuracy보다 F1-score가 더 중요한 경우가 많음.

---

# 7. 생성 태스크 평가

LLM은 생성 모델이므로, 요약·번역·문장생성 같은 생성 태스크에서는 다른 지표가 필요.

대표적으로 다음이 많이 쓰입니다.

- BLEU
- ROUGE
- METEOR
- BERTScore
- Exact Match
- Human Evaluation

---

# 8. BLEU

BLEU는 주로 번역 평가에서 자주 쓰이는 지표.

핵심은 생성문과 기준문(reference) 사이의 n-gram 일치 정도를 보는 것.


$$
\mathrm{BLEU}
=
\mathrm{BP}\cdot
\exp\left(
\sum_{n=1}^{N} w_n \log p_n
\right)
$$

- $\mathrm{BP}$: brevity penalty
- $p_n$: n-gram precision
- $w_n$: 각 n-gram 가중치


## 장점
- 자동 계산 가능
- 번역 비교에 편리

## 한계
- 표현이 달라도 의미가 같으면 낮게 나올 수 있음
- 자유 생성 태스크에는 한계가 큼

---

# 9. ROUGE

ROUGE는 요약 평가에서 자주 쓰임.

대표적으로 ROUGE-N, ROUGE-L이 있다.

## ROUGE-N

reference와 generated text의 n-gram recall 중심 평가.

$$
\mathrm{ROUGE\text{-}N}
=
\frac{
\sum_{\text{gram}_n \in \text{Ref}}
\min(\mathrm{Count}_{\text{gen}}, \mathrm{Count}_{\text{ref}})
}{
\sum_{\text{gram}_n \in \text{Ref}}
\mathrm{Count}_{\text{ref}}
}
$$

## ROUGE-L

최장 공통 부분 수열(Longest Common Subsequence, LCS)을 활용.

$$
\mathrm{ROUGE\text{-}L}
\propto
\frac{\mathrm{LCS}(\text{gen}, \text{ref})}{|\text{ref}|}
$$

## 장점
- 요약 태스크에서 널리 사용
- reference와의 겹침 정도를 직관적으로 볼 수 있음

## 한계
- 의미 보존을 완전히 반영하지 못함
- 창의적 표현에 불리할 수 있음

---

# 10. BERTScore

BLEU나 ROUGE는 문자열 겹침 중심이라 의미적 유사성을 놓칠 수 있다.

이를 보완하려고 embedding 기반 평가가 사용됨.

BERTScore는 생성문과 기준문의 토큰 임베딩 유사도를 활용.

개념적으로는 각 토큰 임베딩 사이의 코사인 유사도를 사용해 precision, recall, F1를 계산.

예를 들어 토큰 임베딩이 $h_i$, $r_j$ 라면

$$
\mathrm{sim}(h_i, r_j)
=
\frac{h_i^\top r_j}{\|h_i\|\|r_j\|}
$$

를 사용함.

그 후 최적 매칭에 가까운 방식으로 score를 계산.

## 장점
- 표면적 단어 일치보다 의미 유사성을 더 잘 반영
- 생성 태스크 평가에 유용

## 한계
- 계산 비용이 큼
- 사용하는 encoder 모델에 따라 결과가 달라질 수 있음

---

# 11. Exact Match (EM)

질의응답이나 수학 문제처럼 정답이 비교적 명확한 경우에는 Exact Match를 씀.

$$
\mathrm{EM}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(\hat{y}_i = y_i)
$$

정답 문자열과 완전히 일치하면 1, 아니면 0.

## 장점
- 해석이 매우 명확함
- QA, closed-form answer 평가에 유리

## 한계
- 표현이 조금만 달라도 0점
- 자유 생성에는 지나치게 엄격

---

# 12. Pass@k

코드 생성이나 수학 문제 해결처럼 여러 개 후보를 생성했을 때 하나라도 맞으면 되는 상황에서는 Pass@k가 중요.

예를 들어 $k$개 후보 중 정답이 하나 이상 있으면 성공으로 보는 지표.


$$
\mathrm{Pass@}k
=
P(\text{top-}k \text{ samples 중 정답 존재})
$$

코드 생성 평가에서는 HumanEval 등에서 자주 사용됨.

생성 샘플 수 $n$, 정답 샘플 수 $c$ 일 때 알려진 unbiased estimator 중 하나는

$$
\mathrm{Pass@}k
=
1 - \frac{\binom{n-c}{k}}{\binom{n}{k}}
$$

## 장점
- 샘플링 기반 생성 능력을 잘 반영
- 코드 생성 평가에 매우 중요

## 한계
- deterministic evaluation과 다름
- 샘플 수와 decoding 전략 영향을 받음

---

# 13. Reasoning 성능 평가

최근 LLM에서는 단순 언어 생성보다 **추론(reasoning)** 능력이 매우 중요.

대표적인 평가 영역은 다음과 같음.

- 수학 문제
- 논리 추론
- 상식 추론
- 다단계 문제 해결

이 경우 보통 벤치마크 정확도를 봄.

예를 들어 데이터셋이 $N$개 문제로 구성되어 있으면

$$
\mathrm{Reasoning\ Accuracy}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(\hat{a}_i = a_i)
$$

로 측정할 수 있음.

대표적인 평가 벤치마크 예시는 다음과 같은 성격을 가짐.

- 일반 지식 및 추론
- 수학 문제 해결
- 상식 추론
- 과학 문제 해결

---

# 14. 코드 생성 평가

LLM의 코드 능력은 별도로 중요하게 평가됩니다.

주요 방식은 다음과 같습니다.

- unit test 통과 여부
- pass@k
- compile 성공 여부
- 실행 결과 정확성

예를 들어 테스트 케이스 집합 $\mathcal{T}$ 에 대해

$$
\mathrm{Code\ Accuracy}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(\text{all tests pass})
$$

로 볼 수 있음.

단순 문자열 일치보다 **실제로 코드가 실행되어 원하는 결과를 내는지**가 더 중요.

---

# 15. 대화 품질 평가

챗봇형 LLM에서는 다음 특성도 중요.

- helpfulness
- harmlessness
- honesty
- coherence
- relevance
- instruction following

이런 항목들은 자동 평가가 어려워서  
사람 평가(human evaluation)나 LLM-as-a-judge 방식도 많이 사용됨.

예를 들어 점수 함수 $s(x, \hat{y})$ 가 있으면

$$
\mathrm{AvgScore}
=
\frac{1}{N}\sum_{i=1}^{N}s(x_i, \hat{y}_i)
$$

처럼 평균 점수로 볼 수 있음.

여기서 $s$는 사람 점수일 수도 있고, 평가용 모델 점수일 수도 있음.

---

# 16. Human Evaluation

가장 직관적인 평가는 사람이 직접 보는 것.

보통 다음 항목을 1점~5점 또는 pairwise preference로 평가함.

- 정확성
- 자연스러움
- 유용성
- 안전성
- 일관성

예를 들어 두 모델 $A$, $B$ 를 비교할 때  
사람이 $A$를 더 선호한 비율을

$$
\mathrm{WinRate}(A,B)
=
\frac{\#(A \succ B)}{\#\text{전체 비교}}
$$

로 나타낼 수 있음.

## 장점
- 실제 사용자 품질에 가까움
- 정교한 평가 가능

## 한계
- 비용이 큼
- 주관성이 개입됨
- 재현성이 낮을 수 있음

---

# 17. LLM-as-a-Judge

최근에는 다른 강한 LLM을 평가자로 사용하기도 함.

예를 들어 입력 $x$, 모델 응답 $y$ 에 대해 평가 모델이 점수 $J(x,y)$ 를 주면

$$
\mathrm{JudgeScore}
=
\frac{1}{N}\sum_{i=1}^{N} J(x_i, y_i)
$$

로 평가할 수 있음.

또는 두 응답 $y_A, y_B$ 중 무엇이 더 좋은지 판정하게 할 수도 있음.

## 장점
- 사람 평가보다 빠름
- 대규모 자동화 가능

## 한계
- 평가 모델의 편향이 들어갈 수 있음
- 특정 스타일을 과도하게 선호할 수 있음

---

# 18. 효율성 평가

LLM은 성능뿐 아니라 비용과 속도도 중요.

대표 지표는 다음과 같습니다.

- latency
- throughput
- memory usage
- FLOPs
- energy cost

## (1) Latency

한 요청을 처리하는 시간.

$$
\mathrm{Latency}
=
t_{\text{end}} - t_{\text{start}}
$$

## (2) Throughput

단위 시간당 처리한 토큰 수.

$$
\mathrm{Throughput}
=
\frac{\text{생성 토큰 수}}{\text{총 시간}}
$$

예를 들어 tokens/sec 형태로 자주 봄.

## (3) Memory

모델 파라미터 수가 $N$, 각 파라미터 비트폭이 $b$ 라면 대략 저장 메모리는

$$
\mathrm{Memory}
\approx \frac{N \cdot b}{8}
\quad \text{bytes}
$$


예를 들어

- FP16: $b=16$
- INT8: $b=8$
- INT4: $b=4$

---

# 19. 안전성 평가

LLM은 성능이 높아도 위험한 응답을 하면 문제가 됨.

그래서 안전성도 중요한 평가 항목.

예를 들어 다음을 봅니다.

- 유해 요청 거부율
- hallucination 비율
- 편향성
- 개인정보 노출 가능성
- jailbreak 저항성

개념적으로 hallucination rate는

$$
\mathrm{Hallucination\ Rate}
=
\frac{\text{사실과 다르거나 근거 없는 응답 수}}{\text{전체 응답 수}}
$$

처럼 볼 수 있음.

안전성 평가는 보통 자동 지표만으로 충분하지 않아 사람 평가와 정책 기반 테스트셋을 함께 사용함.

```python
import time
import math
import torch
import numpy as np

from transformers import AutoTokenizer, AutoModelForCausalLM
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import evaluate
```

```python
# 모델과 토크나이저 준비
model_id = "distilgpt2"

tokenizer = AutoTokenizer.from_pretrained(model_id)
model = AutoModelForCausalLM.from_pretrained(model_id)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = model.to(device)
model.eval()

if tokenizer.pad_token is None:
    tokenizer.pad_token = tokenizer.eos_token
```

Perplexity

$$
P(x_1,\dots,x_T) = \prod_{i=1}^T P(x_t|x_{<t})
$$

언어 모델의 평균의 평균 loss는

$$
ℒ = {\frac{1}{T}} \sum_{i=1}^T \log P(y_t|y_{<t})
$$

이고, perplexity는
$$
PPL = exp(ℒ)
$$

```python
# 평가용 문장 준비
texts = [
    "Quantization reduces memory usage and can speed up inference.",
    "Large language models predict the next token based on previous context.",
    "Perplexity is a common metric for evaluating language modeling quality.",
    "Fine-tuning adapts a pretrained model to a downstream task."
]
```

```python
# 직접 perplxity 계산 함수
def compute_perplexity(model, tokenizer, texts, device):
    encodings = tokenizer(
        texts,
        return_tensors="pt",
        padding=True,
        truncation=True
    )
    input_ids = encodings["input_ids"].to(device)
    attention_mask = encodings["attention_mask"].to(device)

    with torch.no_grad():
        outputs = model(
            input_ids=input_ids,
            attention_mask=attention_mask,
            labels=input_ids
        )
        loss = outputs.loss

    ppl = torch.exp(loss).item()
    return loss.item(), ppl
```

```python
loss, ppl = compute_perplexity(model, tokenizer, texts, device)
print("Cross-Entropy Loss:", loss)
print("Perplexity:", ppl)
```

```python
#문장별 perplexity 보기
def sentence_perplexities(model, tokenizer, texts, device):
    results = []

    for text in texts:
        enc = tokenizer(text, return_tensors="pt")
        input_ids = enc["input_ids"].to(device)

        with torch.no_grad():
            outputs = model(input_ids=input_ids, labels=input_ids)
            loss = outputs.loss.item()
            ppl = math.exp(loss)

        results.append((text, loss, ppl))

    return results
```

```python
results = sentence_perplexities(model, tokenizer, texts, device)

for text, loss_val, ppl_val in results:
    print("=" * 80)
    print("TEXT:", text)
    print("LOSS:", round(loss_val, 4))
    print("PPL :", round(ppl_val, 4))
```


$$
\mathrm{Accuracy}
=
\frac{1}{N}\sum_{i=1}^{N}\mathbf{1}(\hat{y}_i = y_i)
$$

$$
\mathrm{Precision}
=
\frac{TP}{TP + FP}
$$


$$
\mathrm{Recall}
=
\frac{TP}{TP + FN}
$$


$$
F1
=
\frac{2 \cdot \mathrm{Precision} \cdot \mathrm{Recall}}
{\mathrm{Precision} + \mathrm{Recall}}
$$


- $TP$: True Positive
- $FP$: False Positive
- $FN$: False Negative

```python
# 예제 정답과 예측값 준비
y_true = [1, 0, 1, 1, 0, 0, 1, 0]
y_pred = [1, 0, 1, 0, 0, 0, 1, 1]
```

```python
# 지표 계산
acc = accuracy_score(y_true, y_pred)
prec = precision_score(y_true, y_pred)
rec = recall_score(y_true, y_pred)
f1 = f1_score(y_true, y_pred)

print("Accuracy :", acc)
print("Precision:", prec)
print("Recall   :", rec)
print("F1-score :", f1)
```

```python
# 직접 수식
y_true_np = np.array(y_true)
y_pred_np = np.array(y_pred)

TP = np.sum((y_true_np == 1) & (y_pred_np == 1))
FP = np.sum((y_true_np == 0) & (y_pred_np == 1))
FN = np.sum((y_true_np == 1) & (y_pred_np == 0))
TN = np.sum((y_true_np == 0) & (y_pred_np == 0))

print("TP:", TP, "FP:", FP, "FN:", FN, "TN:", TN)

precision_manual = TP / (TP + FP)
recall_manual = TP / (TP + FN)
f1_manual = 2 * precision_manual * recall_manual / (precision_manual + recall_manual)

print("Manual Precision:", precision_manual)
print("Manual Recall   :", recall_manual)
print("Manual F1       :", f1_manual)
```

\mathrm{BLEU}
=
\mathrm{BP}\cdot
\exp\left(
\sum_{n=1}^{N} w_n \log p_n
\right)
$$

- $\mathrm{BP}$: brevity penalty
- $p_n$: n-gram precision
- $w_n$: 각 n-gram 가중치

ROUGE-N

reference와 generated text의 n-gram recall 중심 평가.

$$
\mathrm{ROUGE\text{-}N}
=
\frac{
\sum_{\text{gram}_n \in \text{Ref}}
\min(\mathrm{Count}_{\text{gen}}, \mathrm{Count}_{\text{ref}})
}{
\sum_{\text{gram}_n \in \text{Ref}}
\mathrm{Count}_{\text{ref}}
}
$$
ROUGE-L

최장 공통 부분 수열(Longest Common Subsequence, LCS)을 활용.

$$
\mathrm{ROUGE\text{-}L}
\propto
\frac{\mathrm{LCS}(\text{gen}, \text{ref})}{|\text{ref}|}
$$

```python
def generate_response_only(model, tokenizer, prompt, device, max_new_tokens=40):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    input_len = inputs["input_ids"].shape[1]

    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=True,
            top_k=50,
            top_p=0.95,
            temperature=0.8,
            pad_token_id=tokenizer.eos_token_id
        )

    generated_ids = outputs[0][input_len:]
    response = tokenizer.decode(generated_ids, skip_special_tokens=True).strip()
    return response
```

- 생성 속도 평가

Latency

$$
\mathrm{Latency}
=
t_{\text{end}} - t_{\text{start}}
$$

Throughput

$$
\mathrm{Throughput}
=
\frac{\text{생성 토큰 수}}{\text{총 시간}}
$$

```python
# latency 측정 함수
def benchmark_generation(model, tokenizer, prompt, device, max_new_tokens=50):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)

    start = time.time()
    with torch.no_grad():
        outputs = model.generate(
            **inputs,
            max_new_tokens=max_new_tokens,
            do_sample=False,
            pad_token_id=tokenizer.eos_token_id
        )
    end = time.time()

    generated_ids = outputs[0]
    input_len = inputs["input_ids"].shape[1]
    total_len = generated_ids.shape[0] if generated_ids.ndim == 1 else generated_ids.shape[1]
    new_tokens = total_len - input_len

    latency = end - start
    throughput = new_tokens / latency if latency > 0 else float("inf")

    text = tokenizer.decode(outputs[0], skip_special_tokens=True)
    return latency, throughput, new_tokens, text
```

```python
latency, throughput, new_tokens, text = benchmark_generation(
    model,
    tokenizer,
    "Explain weight-only quantization simply.",
    device,
    max_new_tokens=50
)

print("Latency (sec):", latency)          # latency 실제 배포하면 오래걸림 값을 줄이는 게 좋다
print("Generated tokens:", new_tokens)
print("Throughput (tokens/sec):", throughput)
print(text)
```

```python
# 여러 번 평균 내기
def repeat_benchmark(model, tokenizer, prompt, device, n_runs=5, max_new_tokens=50):
    latencies = []
    throughputs = []

    for _ in range(n_runs):
        latency, throughput, _, _ = benchmark_generation(
            model, tokenizer, prompt, device, max_new_tokens=max_new_tokens
        )
        latencies.append(latency)
        throughputs.append(throughput)

    return np.mean(latencies), np.std(latencies), np.mean(throughputs), np.std(throughputs)
```

```python
mean_lat, std_lat, mean_tp, std_tp = repeat_benchmark(
    model,
    tokenizer,
    "Explain quantization in one paragraph.",
    device,
    n_runs=5,
    max_new_tokens=60
)

print("Mean latency   :", mean_lat)
print("Std latency    :", std_lat)
print("Mean throughput:", mean_tp)
print("Std throughput :", std_tp)
```
